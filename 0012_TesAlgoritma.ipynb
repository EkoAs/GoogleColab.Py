```python
a = [8,2,3,4,5,1,2,3,4,5]
target=8
total =0
for i in range(len(a)-2):
  j=i+1
  for j in range(len(a)-1):
    total+=1
    if a[i]==target:
      print(f"\ntrue")
      print(f"loop ke {i} : {a[i]} == {a[j]} : Total Loops {total}")
      break
    else:
      print('\nfalse')
      print(f"loop ke {i} : {a[i]}=={a[j]}  : Total Loops {total}")
      break
```

```python
a = [8, 2, 3, 4, 5, 1, 2, 3, 4, 5]
n = len(a)
total = 0

def cek_duplikat(a):
    global total
    n = len(a)
    for i in range(n - 1):            # i = 0 .. n-2
        for j in range(i + 1, n):     # j = i+1 .. n-1
            total += 1                # hitung Basic Operation
            if a[i] == a[j]:
                print(f"true: a[{i}] == a[{j}] = {a[i]}, total perbandingan = {total}")
                return True
    print(f"false, total perbandingan = {total}")
    return False

cek_duplikat(a)
```

```python
a = [1, 2, 3, 4, 5]
target = 5

# Mencari semua pasangan yang jumlahnya sama dengan target
hasil = []
for i in range(len(a)):
    for j in range(i + 1, len(a)):
        if a[i] + a[j] == target:
            hasil.append((a[i], a[j]))

print(f"Pasangan yang menghasilkan jumlah {target}: {hasil}")
```
