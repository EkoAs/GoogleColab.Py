```python
a = [8,2,3,4,5,1,2,3,4,5]
target=8
for i in range(len(a)-2):
  j=i+1
  for j in range(len(a)-1):
    if a[i]==target:
      print(f"\ntrue")
      print(f"loop ke {i} : {a[i]} == {a[j]}")
      break
    else:
      print('\nfalse')
      print(f"loop ke {i} : {a[i]}=={a[j]}")
      break
```
